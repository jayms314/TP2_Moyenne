```python
coefficients = [5, 6, 8, 7, 8]
notes = []

total = 0
total_coefficients = 0

for i in range(17):
    Ressources= i+101
    N = float(input("Note de la ressource R ",Ressources))
    notes.append(N)

    total = total + N * coefficients[i]
    total_coefficients = total_coefficients + coefficients[i]

moyenne = total / total_coefficients

print("Notes :", notes)
print("Coefficients :", coefficients)
print("Votre moyenne :", moyenne)

```
